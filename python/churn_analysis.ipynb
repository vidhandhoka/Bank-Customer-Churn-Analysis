```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

Import Data

```python
df = pd.read_csv('BankChurners.csv')
df.head()
```

Dataset Understanding

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.drop(columns=[
    "CLIENTNUM",
    "Naive_Bayes_Classifier_Attrition_Flag_Card_Category_Contacts_Count_12_mon_Dependent_count_Education_Level_Months_Inactive_12_mon_1",
    "Naive_Bayes_Classifier_Attrition_Flag_Card_Category_Contacts_Count_12_mon_Dependent_count_Education_Level_Months_Inactive_12_mon_2"
], inplace=True)
```

```python
df.describe()
```

```python
df.columns
```

```python
df.rename(
    columns={
        "Customer_Age": "Age",
        "Income_Category": "Income_Group",
        "Total_Trans_Ct": "Transaction_Count"
    },
    inplace=True
)
```

```python
df["Gender"]=df["Gender"].replace({"M": "Male", "F": "Female" })
```

```python
df["Gender"].value_counts()
df
```

```python
df.loc[df["Age"] < 30, "Age_Group"] = "Young"
df.loc[df["Age"] >= 30, "Age_Group"] = "Adult"
df.loc[df["Age"] >= 60, "Age_Group"] = "Senior"
```

```python
df["Churn_Flag"] = 0
df.loc[
    df["Attrition_Flag"] == "Attrited Customer",
    "Churn_Flag"
] = 1;

df["Churn_Flag"].value_counts()
```

```python
df["Attrition_Flag"] = df["Attrition_Flag"].str.title()
```

export clean dataset

```python
df.to_csv(
    "BankChurners_Cleaned.csv",
    index=False
)
```

Find the top 10 churned customers by transaction amount.

```python
top_churned = (
    df[df["Attrition_Flag"] == "Attrited Customer"]
    .sort_values("Total_Trans_Amt", ascending=False)
    .head(10)
)
top_churned
```

Income Group vs Churn

```python
income_churn = (
    df.groupby("Income_Group")["Churn_Flag"]
    .mean()
    .mul(100)
    .sort_values(ascending=False)
)

income_churn
```

Age group vs Churn

```python
age_churn = (
    df.groupby("Age_Group")["Churn_Flag"]
    .mean()
    .mul(100)
    .sort_values(ascending=False)
)

age_churn
```

Does the type of credit card a customer holds relate to their churn rate?

```python
card_churn = (
    df.groupby("Card_Category")["Churn_Flag"]
    .mean()
    .mul(100)
    .sort_values(ascending=False)
)

card_churn
```

Do churned customers have different transaction activity compared with existing customers?

```python
transaction_churn = (
    df.groupby("Attrition_Flag")["Transaction_Count"]
    .agg(["mean", "median", "min", "max"])
)

transaction_churn
```

Do churned customers have different transaction spending behavior?

```python
amount_churn = (
    df.groupby("Attrition_Flag")["Total_Trans_Amt"]
    .agg(["mean", "median", "min", "max"])
)

amount_churn
```

Does customer inactivity relate to churn?

```python
inactive_churn = (
    df.groupby("Months_Inactive_12_mon")["Churn_Flag"]
    .agg(["count", "sum", "mean"])
)

inactive_churn["Churn_Rate_%"] = inactive_churn["mean"] * 100

inactive_churn
```

Does credit card utilization differ between existing and churned customers?


```python
utilization_churn = (
    df.groupby("Attrition_Flag")["Avg_Utilization_Ratio"]
    .agg(["mean", "median", "min", "max"])
)

utilization_churn
```

Does the number of products/relationships a customer has with the bank relate to churn?

```python
relationship_churn = (
    df.groupby("Total_Relationship_Count")["Churn_Flag"]
    .agg(["count", "sum", "mean"])
)

relationship_churn["Churn_Rate_%"] = (
    relationship_churn["mean"] * 100
)

relationship_churn
```

Does a change in transaction activity relate to customer churn?

```python
transaction_change_churn = (
    df.groupby("Attrition_Flag")["Total_Ct_Chng_Q4_Q1"]
    .agg(["count", "mean", "median", "min", "max"])
)

transaction_change_churn
```

Does a change in transaction amount differ between existing and churned customers?

```python
amount_change_churn = (
    df.groupby("Attrition_Flag")["Total_Amt_Chng_Q4_Q1"]
    .agg(["count", "mean", "median", "min", "max"])
)

amount_change_churn
```

Which numerical variables have the strongest relationship with customer churn?

```python
numeric_cols = [
    "Age",
    "Credit_Limit",
    "Total_Revolving_Bal",
    "Avg_Open_To_Buy",
    "Total_Trans_Amt",
    "Transaction_Count",
    "Total_Ct_Chng_Q4_Q1",
    "Total_Amt_Chng_Q4_Q1",
    "Total_Relationship_Count",
    "Months_Inactive_12_mon",
    "Contacts_Count_12_mon",
    "Avg_Utilization_Ratio",
    "Churn_Flag"
]

corr = df[numeric_cols].corr()

churn_corr = (
    corr["Churn_Flag"]
    .drop("Churn_Flag")
    .sort_values()
)

churn_corr
```
